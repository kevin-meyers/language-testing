```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
df = pd.read_csv('/home/kevin/Downloads/Cookie Run Packages - Sheet1.csv')
```

```python
prepared_df = df.fillna(0).replace(True, 1).drop('Date', axis=1).drop('Name', axis=1)
```

```python
lr = LinearRegression()
lr.fit(X=prepared_df.drop('Price', axis=1), y=prepared_df['Price'])
```

```python
final_df = df[['Name', 'Price']]
final_df['Value'] = lr.predict(prepared_df.drop('Price', axis=1)) / prepared_df['Price']
```

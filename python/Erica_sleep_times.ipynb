```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/home/kevin/Downloads/AutoSleep-20200125-to-20201022.csv')
df = df[df['bedtime'].notna()]
df = df[df['deep'].notna()]
df = df[df['asleep'].notna()]
```

```python
def hours_to_double(h_string):
    return float(h_string[:2] + '.' + h_string[3:5] + h_string[6:])
```

```python
df['bedtime'] = df['bedtime'].apply(lambda x: hours_to_double(x[11:]))
df['deep'] = df['deep'].apply(hours_to_double)
df['asleep'] = df['asleep'].apply(hours_to_double)
df['inBed'] = df['inBed'].apply(hours_to_double)
```

```python
df['deep_to_total_ratio'] = df['deep'] / df['asleep']
```

```python
import pandas as pd
```

```python
df = pd.read_csv('/home/kevin/Downloads/Rapid HIV and HCV Test Log .xlsx - Sheet2.csv', skiprows=7)
```

```python
required = 'Control Test Required?                                                                                      (if yes, list reason) '
```

```python
count = '                - After incident    Inventory  test kit count    '
```

```python
filtered_df = df.loc[~(df[required].str.contains('Yes') & df[required].str.contains('> 7') & ~df[required].str.contains('new'))]
```

```python
unfiltered_df = df.loc[(df[required].str.contains('Yes') & df[required].str.contains('> 7') & ~df[required].str.contains('new'))]
```

```python
unfiltered_df['*Control test Incubation     START: Times-'] = unfiltered_df['*Control test Incubation     START: Times-'].apply(lambda x: x.split('- ')[0] + '-')
```

```python
filtered_df['diff'] = 1
```

```python
filtered_df.loc[filtered_df[required].str.contains('HIV') & filtered_df[required].str.contains('Yes'), 'diff'] = 4
```

```python
filtered_df.loc[filtered_df[required].str.contains('HCV') & filtered_df[required].str.contains('Yes'), 'diff'] = 3
```

```python
hiv_count = 25
hcv_count = 25
new_counts = []
for (i, (_, r, d)) in enumerate(filtered_df[[count, 'diff']].itertuples()):
    if i < 59:
        new_counts.append(r)
        continue
    if 'HCV' in r:
        hcv_count = (hcv_count - d + 25) % 25
        print(hcv_count)
        print(hcv_count)
        new_counts.append(f'HCV= {hcv_count}')
    elif 'HIV' in r:
        hiv_count = (hiv_count - d + 25) % 25
        new_counts.append(f'HIV= {hiv_count}')
```

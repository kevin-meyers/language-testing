```python
import pandas as pd
```

```python
df = pd.read_csv('EURUSD-2020-11.csv', names=['pid', 'datetime', 'low', 'high'])
```

```python
df.head()
```

```python
len(df)
```

```python
df['datetime']  = pd.to_datetime(df['datetime'])
```

```python
hourly_df = df.groupby(pd.Grouper(key='datetime', freq='1h')).mean()
```

```python
hourly_df.plot()
```
